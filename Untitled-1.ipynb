```python
# Importer le DataSet depuis csv
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sklearn as sklearn
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import OneHotEncoder


df = pd.read_csv('BMW sales data (2010-2024) (1) (1).csv')
```

```python
# Afficher les 10 premières lignes du DataFrame
print(df.head(10))

# Afficher les informations sur le DataFrame
print(df.info())

# Afficher les graphiques statistiques descriptives du DataFrame
print(df.describe())
```

```python
# Afficher les boxplots pour chaque colonne numérique
for column in df.select_dtypes(include=[np.number]).columns:
    plt.figure(figsize=(8, 4))
    sns.boxplot(x=df[column])
    plt.title(f'Boxplot of {column}')
    plt.show()
    
# Afficher les histogrammes pour chaque colonne numérique
for column in df.select_dtypes(include=[np.number]).columns:
    plt.figure(figsize=(8, 4))
    sns.histplot(df[column], kde=True)
    plt.title(f'Histogram of {column}')
    plt.show()
```
